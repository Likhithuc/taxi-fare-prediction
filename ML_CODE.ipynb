```python
import pandas as pd
import random
```

```python
def generate_taxi(num_samples=1000):
  data=[]
  for _ in range(num_samples):
    distance=round(random.uniform(1,20),2)
    passenger_count=random.randint(1,4)
    base_fare=3.0
    fare=base_fare+(distance*0.5)+(passenger_count*0.5)+random.uniform(-1,1)
    data.append([distance,passenger_count,round(fare,2)])
  df=pd.DataFrame(data,columns=["distance","passenger_count","fare_amount"])
  return df
dataset=generate_taxi(5000)
print(dataset.head())

dataset.to_csv("taxi_dataset.csv",index=False)

print("✅ Synthetic dataset saved as synthetic_taxi_fare.csv")
```

```python
import pandas as pd
import random
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error,r2_score
import joblib
```

```python
x=dataset[["distance","passenger_count"]]
y=dataset["fare_amount"]
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model=RandomForestRegressor(n_estimators=100,random_state=42)
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
```

```python
print(X_train.head())
```

```python
#Model selection by Score
mse=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
print(f"Mean Squared Error:{mse:.2f}")
print("R-squared:",r2)
```

```python
joblib.dump(model,"taxi_fare_model.pkl")
print("Model saved taxi_fare_model.pkl")
```

```python
loaded_model=joblib.load("taxi_fare_model.pkl")
new_trip=pd.DataFrame({'distance':[12.5],'passenger_count':[2]})
predicted_fare=loaded_model.predict(new_trip)
print(f"Predicted fare amount:{predicted_fare[0]:.2f}")
```
